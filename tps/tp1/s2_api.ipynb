## TP1 - API choisie : Wikipédia



```python
headers = {'User-Agent': 'TAC-ULB-TP1/1.0 (etudiant ULB)'}

def get_summary(term, lang="fr"):
    url = f"https://{lang}.wikipedia.org/api/rest_v1/page/summary/{term.replace(' ', '_')}"
    resp = requests.get(url, headers=headers)
    if resp.status_code != 200:
        return None
    data = resp.json()
    return {
        "titre": data["title"],
        "description": data.get("description"),
        "resume": data["extract"],
    }
```

```python
entities = ["Léopold II (roi des Belges)", "Anvers", "Charleroi", "Adolphe Max"]

for entity in entities:
    result = get_summary(entity)
    if result is None:
        print(f"{entity}: introuvable")
        continue
    print(f"{result['titre']} - {result['description']}")
    print(result["resume"][:200] + "...")
    print("---")
```
